**NUMBER GUESSING GAME**

Problem Statement: Create a Python program that implements a simple number
guessing game using a while loop. The program should make use of control
statements such as else, break, and continue.

```python
'''1. Generate a random number between 1 and 10 that the user has to guess.
Import random and use randint function'''

import random
# Generate random number between 1 and 10
Secret_number = random.randint(1, 10)
```

```python
#Asking the user to guess the number


Attempts=3   # max number of aattempts for thw user
while Attempts>0:
    Guess_number=int(input("Guess the number between 1 and 10: "))

#check if guess is out of range

    if Guess_number<0 and Guess_number>10:
      print("Oops,the guess is out of range")
      continue   #rest of the loop will not work

#checking the guess
    if Guess_number<Secret_number :
      print("Too low,try again")

    elif Guess_number>Secret_number :
      print("Too high,try again")

    else :
      print("Congratulations!,You guessed right")
      break   # to exit the loop

# decreasing the number of attempts
    Attempts=Attempts-1

else :
  print("Out of attempts,Better luck next time!")  #  if the user runs out of attempts without guessing the correct number

```

**FOR LOOP**

Problem Statement: Create a Python program that generates and prints a
multiplication table (from 1 to 10) for a given number using a for loop and the range
function.

```python
#taking the input from user to get table

Multi_numb=int(input("Enter the number to generate multiplication table :"))

#generating the table

for i in range(1,11):
  print( Multi_numb,"*",i,"=",Multi_numb*i)
```

**FUNCTION**

Problem Statement: Create a Python program that calculates the Body Mass Index
(BMI).
Hint: BMI = weight (kg) / [height (m)]²  

```python
# defining a function

def cal_bmi(weight,height) :

  if height > 100:        # Assuming if height is greater than 100, it's in cm
    height = height / 100 # Convert to meters

  BMI = weight / (height**2)
  return BMI

# entering the data by user

weight=float(input("Enter your weight in kg : "))
height=float(input("Enter your height in m (or cm if > 100): "))

#calculating BMI

bmi=cal_bmi(weight,height)
print("Your BMI is : ", round(bmi, 2))
```
