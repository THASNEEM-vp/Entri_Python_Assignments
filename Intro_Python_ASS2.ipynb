**LIST (CREATION,MODIFICATION,ACCESS)**

**1.List creation**


```python
age_list=[24,25,26,27,28]                       # list 1 creation
name_list=["Anna","Sam","Tom","Mary","Manu"]    # list 2 created

print("Age_list : ",age_list)
print("Name_list : ",name_list)
```

**2.List Operations/Modifications**

```python
name_list.append("Yazhini")        #appended "Yazhini" to the existing list
print("Name list : ",name_list)
```

```python
print("Age list :",age_list)
age_list.insert(2,30)             #insert value 30 to index 2
print("Age list :",age_list)
```

```python
print("Name_list : ",name_list)
name_list.remove("Yazhini")        #removed "Yazhini" from the list
print("Name_list : ",name_list)
```

```python
age_list.pop()                  #removed the last element
print("Age_list : ",age_list)
```

```python
print("Age list : ",age_list)
age_list.extend([29,30,26])     # added values 29,30,26 to existing age_list
print("Age_New : ",age_list)
```

```python
print("Age list : ",age_list)
age_list.sort(reverse=True)     #sorted the age_list in descending order
print("Age list : ",age_list)
```

```python
print("Age list : ",age_list)
print("Maximum age : ",max(age_list))     #maximum age in the age list
print("Minimum age : ",min(age_list))     #minimum age in the age list
print("Sum of ages is : ",sum(age_list))  #sum of all ages in the list
```

**Accessing list element**

```python
print("Name list : ",name_list)
print("First element : ",name_list[0])   # gives first element in name list
print("Last_element : ",name_list[-1])   # gives last element in the name list
print("Elements from index 2 to 4 are : ",name_list[2:5]) #elements from index 2 to 4
print("Reverse : ",name_list[::-1])      #print the list in reverse order
```

**Dictionary**

```python
#dictioinary creation

student_marks={"Anna":90,"Ram":87,"Mary":100,"Lina":50,"Polo":75}
print("Dictionary : ",student_marks)
print("Values : ",student_marks.values())
print("Keys : ",student_marks.keys())
```

```python
#accessing the mark of "Lina" and printing it

print("Dictionary : ",student_marks)
print("Mark of student Lina : ",student_marks["Lina"])
```

```python
# Adding a new student "Janani" with a mark of 80 to the student_marks dictionary

student_marks["Janani"]=80
print("Dictionary : ",student_marks)
```

```python
# Updating the mark of any one older student to 82

print("student marks : ",student_marks)
student_marks["Ram"]=82
print("New student marks : ",student_marks)
```

```python
''' using the keys(), values(), and items() methods to print all keys, values, and
key-value pairs in the student_marks dictionary'''

print("Keys : ",student_marks.keys())
print("Values : ",student_marks.values())
print("Items : ",student_marks.items())
```

**Set operations**

```python
# Creating a set called my_set

my_set={"a","e","i","o","u","a","a","i"}
print(my_set)
print(type(my_set))

'''since set is a unordered and mutable collection of data where it doesn't support
   duplicate elements the duplicate elements are rejected in the output'''
```

```python
'''Attempting to change the value of my_set[4] = 's'. If code throws an error, provide
   an explanation'''

print(my_set)
my_set[4]="s"
print(my_set)
```

Explaination :Since sets are mutable they will not allow replcing a value in particular index because they dont have proper indexing but they support addding and removing elements

```python
# Create two sets

set1={1,3,5,7,9}
set2={2,3,5,8,10}

print("Set 1 : ",set1)
print("Set 2 : ",set2)
```

```python
# Compute and print the union and intersection of set1 and set2

print("Union if set1 and set2 is : ",set1.union(set2))
print("Intersection of set1 and set2 is : ",set1.intersection(set2))
```

**Operators & Conditional statements -(IF,ELIF,ELSE)**

```python
# Performance Category Program:

# 1.Prompt user for Input. Score range should be from 0 to 10 (both inclusive)

score=float(input("Enter the score(0-10) : "))

'''Find the performance category based on the input score using following criteria:
a. Above Average:  Score greater than 7
b. Average: Score between 4 and 7(both inclusive)
c. Below Average: Score lesser than 4'''

if score<0 or score>10 :
  print("Invalid Entry")
elif score>7 :
  print("Score : ",score)
  print("Above Average")
elif score>=4 and score<=7 :
        print("Score : ",score)
        print("Average")
else :
  print("Score : ",score)
  print("Below Average")
```

```python
# Output: Print the Performance category

score=float(input("Enter the score(0-10) : "))
if score<0 or score>10 :
  category="Invalid Entry"
elif score>7 :
  print("Score : ",score)
  category="Above Average"
elif score>=4 and score<=7 :
        print("Score : ",score)
        category="Average"
else :
  print("Score : ",score)
  category="Below Average"
print("Performance Category :",category)
```
