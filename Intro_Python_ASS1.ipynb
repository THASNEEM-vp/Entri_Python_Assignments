# **2.String Slicing and Indexing**

```python
# printing the first character of the string using index accessing

print(result)
print("Output : ",result[0])
```

```python
# Printing the last character of the string

print(result)
print("Output : ",result[-1])
```

```python
# print the first 5 characters of the string

print(result)
print("Output : ",result[0:5])  #slicing
```

```python
# Print the last 11 characters of the string

print(result)
print("Output : ",result[-11:])   #negative indexing
```

```python
# Print the string in reverse

print(result)
print("Output : ",result[::-1])
```

```python
# Use slicing and print the word “Python” from the existing string

print(result)
print(result[27:34])
```

# **3.String Methods**

```python
strM="Python beginner tutorial"
print("Output : ",strM)
print(type(strM))

# Convert the sentence to uppercase

print("Output : ",strM.upper())
```

```python
# Convert the sentence to lowercase

print(strM)
print("Output : ",strM.lower())
```

```python
# Use Capitalize and return the sentence to the original input form

print(strM)
print("Output :",strM.capitalize())
```

```python
# Count the total number of occurrences of character ‘t’  in the string

print(strM)
print("Count : ",strM.count('t'))
```

```python
''' Replace all occurrences of “Python” with “Machine Learning”  in the input string
strM = “Python beginner tutorial” '''

print(strM)
print("Output : ",strM.replace("Python","Machine Learning"))
```

# **Tuples**

```python
# tuple creation

tup1=(10,20,30)
tup2=(40,50,60)
print(tup1)
print(tup2)
print(type(tup1))
print(type(tup2))
```

```python
# Concatenate the two tuples and store it in “t_combine”

t_combine=tup1+tup2
print("Output : ",t_combine)
```

```python
# Repeat the elements of “t_combine” 3 times

print(t_combine)
print("Output : ",t_combine*3)
```

```python
# Access the 3rd element from “t_combine”

print(t_combine)
print("Output : ",t_combine[2])
```

```python
# Access the first three elements from “t_combine”

print(t_combine)
print("Output : ",t_combine[:3])
```

```python
# Access the last three elements from “t_combine”

print(t_combine)
print("Output : ",t_combine[-3:])
```
